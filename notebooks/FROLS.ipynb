# FROLS - Forward Regression Orthogonal Least Squares

Non-linear system identification using Gram-Schmidt orthogonalization with ERR-based selection.
Rust backend for computation, Python for plotting.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
from src import Frols
```

## Data from a flexible robot arm

```python
robot_arm = pd.read_csv('../data/robot-arm.csv')
u = robot_arm['u'].values
y = robot_arm['y'].values

frols_robot = Frols(u, y, nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10)
frols_robot.run(validation=True, title='Robot Arm')
```
