# Non-Linear System Identification — Benchmark Completo

Executa os 3 algoritmos (SEMP, FROLS, Gram-Schmidt) em todos os 5 datasets e compara os resultados.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
from src import Semp, Frols, GramSchmidt
```

```python
datasets = {
    'Robot Arm': {
        'file': '../data/robot-arm.csv', 'u_col': 'u', 'y_col': 'y',
        'semp':  dict(l=1, nu=1, ny=12, ne=0),
        'frols': dict(nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10),
        'gs':    dict(nu=2, ny=2, nl=3, n_theta=4),
    },
    'Tanque': {
        'file': '../data/tanque.csv', 'u_col': 'uEst', 'y_col': 'yEst',
        'semp':  dict(l=1, nu=2, ny=3, ne=0),
        'frols': dict(nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10),
        'gs':    dict(nu=2, ny=3, nl=3, n_theta=4),
    },
    'Silver Box': {
        'file': '../data/SNLS80mV.csv', 'u_col': 'V1', 'y_col': 'V2',
        'semp':  dict(l=1, nu=9, ny=5, ne=0),
        'frols': dict(nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10),
        'gs':    dict(nu=2, ny=2, nl=3, n_theta=4),
    },
    'Ball and Beam': {
        'file': '../data/ball-and-beam.csv', 'u_col': 'u', 'y_col': 'y',
        'semp':  dict(l=1, nu=4, ny=5, ne=0),
        'frols': dict(nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10),
        'gs':    dict(nu=2, ny=2, nl=3, n_theta=4),
    },
    'Exchanger': {
        'file': '../data/exchanger.csv', 'u_col': 'q', 'y_col': 'th',
        'semp':  dict(l=1, nu=2, ny=2, ne=0),
        'frols': dict(nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10),
        'gs':    dict(nu=2, ny=2, nl=3, n_theta=4),
    },
    'Open Loop': {
        'file': '../data/dados_open_loop.csv', 'u_col': 'open_duty', 'y_col': 'y',
        'semp':  dict(l=1, nu=2, ny=2, ne=0),
        'frols': dict(nu=5, ny=5, ne=0, nl=1, tol=0.0, max_iter=10),
        'gs':    dict(nu=2, ny=2, nl=3, n_theta=4),
    },
}
```

```python
results = []

for name, cfg in datasets.items():
    print(f"\n{'#' * 60}")
    print(f"  Dataset: {name}")
    print(f"{'#' * 60}\n")

    df = pd.read_csv(cfg['file'])
    u = df[cfg['u_col']].values
    y = df[cfg['y_col']].values

    # ── SEMP ──
    print(f"--- SEMP ---")
    semp = Semp(u, y, **cfg['semp'])
    mse_semp = semp.run(validation=True, title=f'{name} — SEMP')

    # ── FROLS ──
    print(f"\n--- FROLS ---")
    frols = Frols(u, y, **cfg['frols'])
    mse_frols = frols.run(validation=True, title=f'{name} — FROLS')

    # ── Gram-Schmidt ──
    print(f"\n--- Gram-Schmidt ---")
    gs = GramSchmidt(u, y, **cfg['gs'])
    mse_gs = gs.run(validation=True, title=f'{name} — Gram-Schmidt')

    results.append({
        'Dataset': name,
        'SEMP (MSE test)': mse_semp,
        'FROLS (MSE test)': mse_frols,
        'Gram-Schmidt (MSE test)': mse_gs,
    })
```

## Tabela comparativa

```python
summary = pd.DataFrame(results).set_index('Dataset')
summary.style.format('{:.6e}').highlight_min(axis=1, color='green')
```
