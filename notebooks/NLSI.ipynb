# Gram-Schmidt Structure Selection

Non-linear system identification with Gram-Schmidt orthogonalization using Rust backend.
Compares full model vs selected model (ERR-based structure selection).

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
from src import GramSchmidt
```

## Data from a flexible robot arm

```python
robot_arm = pd.read_csv('../data/robot-arm.csv')
u = robot_arm['u'].values
y = robot_arm['y'].values

gs_robot = GramSchmidt(u, y, nu=2, ny=2, nl=3, n_theta=4)
gs_robot.run(validation=True, title='Robot Arm')
```
